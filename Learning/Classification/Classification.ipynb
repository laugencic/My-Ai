#types of learners
lazy learners>> they store the training data and wait until a testind data appears
    .they have more prediction time 
eager learners>> they construct a classification model based on given training data before getting data for predictions

#classification algorithms 
logistic regression> uses one or more independent variables to determine an outcome                                                                                          
    .the outcome will only have two possible outcomes i.e true or false
    .the main goal is best fitting rel btwn dependent variable and i.v
    .it specifies how a set of i.v affect the d.v
    .only works when the predicted variable is binary
    .assumes data has no missing values and the values are independent of each other 

Naive bayes>an algorithm based on bayes theorem, it assumes that the presence of a feature in a clas is urelated to the presence of any other feature
    .is to make and useful for large datasets
    .requires a small training data
    . fast
    .they are poor estimators

    uses:
    .spam filters, disease prediction, document classification
Stochastic gradient descent>> an effective and simple approach to fit linear models
    >it supports different loss functions and penalties for classification
    >it invloves calculating the derivative from each training data instance and calculate the update immediately
    .ease of use and efficiency
    .requires a number of hyperpcarameters and very sensitive to feature scaling


K-nearest neighbour>a lazy learning algorithm that stores all training data in n-dimensional space
    .It is a lazy learning algorithm as it does not focus on constructing a general internal model, instead it works on storing instances of training data
    .it takes a label points and uses them to label other points
    .ease of of use and very robust to noisy training data 
    .the is no need to determine the value of k and computation cost is high
    .uses:handwriting detection, image&video recognition, stock analysis

Decion tree>> builts the classification model in the form of a tree structure. It utilizes the if-then rules which are equally exhaustive and mutually exclusive
    .made up of nodes and trees
    .the tree is constructed in a top down recursive method
    .a leaf contains a classificationa or decision
    .handles categorical and numerical data
    .requires little data preparation
    .can create complex trees which may not serve the intended purpose
    uses: data exploration, pattern recognition


Random forest/random decison trees>> it operates by constructing a multitude of decision treest at training tim and outputs the class that is the mode of the classes or classification of mean prediction of individual trees
    .fits a number of trees on various subsamples of data and uses an average to increase the accuracy
    .it is more accurate than decision tree
    .complex in implimentation and slow in real time prediction


Artificial neural network>> a neural network consists of neurons that are arranged in layers.
They take some input vector and convert it into an output.The process involves each neuron taking input and applying a function which is often a non-linear function to it and then passes the output to the next layer
    .has a high tolerance to noisy data , able to classify and train patterns and it performs better with contious value inputs
    .has poor interpretation

    .uses: coloration of black and white images , computer vision , photos bases on facial features


Support vector machine>> this a classifier that represents the training data as points in spave separated by categories by a gap as wide as possible.New points are then added to spave by predicting which category they fall into and which space they belong to
    .uses a subset of training points in the decion function which makes it memory efficient and vey effective in high dimensional spaces
    .it doesn't directly provide probability estimates 

Classifier Evaluation:> this involves determing how good my model is at making the correct predictions

Holdout method>in this method, data is set into two parts test and train 

Cross validation> in this, data is randombly partitioned into k mutually exclusive sebsets , each of the same size. One of these is then kept for training the model

ROC curve>>used for visual comparison of classification models, which shows the relationship between the true positive rate and the false positive rate. The area under the roc corve is the measure of the accuracy of the model

Classification report >>it summarizes how well your model performed, broken down into :precion, recall, f1-score, support

precion> When the model says this class, how often is it right
    .High precision = fewer false alarms

Recall>Out of all real cases, how many did we catch?
    .High recall = fewer misses

F1-score>Balance between precision and recall
Support> How many actual samples exist for this class
